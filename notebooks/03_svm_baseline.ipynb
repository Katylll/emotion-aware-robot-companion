```python
import os
import pandas as pd
import numpy as np

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
PROJECT_ROOT = ".."
RAW_DATA_PATH = os.path.join(PROJECT_ROOT, "data", "raw", "fer2013.csv")

df = pd.read_csv(RAW_DATA_PATH)

EMOTIONS = {
    0: "Angry",
    1: "Disgust",
    2: "Fear",
    3: "Happy",
    4: "Sad",
    5: "Surprise",
    6: "Neutral"
}
```

```python
train_df = df[df["Usage"] == "Training"]
val_df = df[df["Usage"] == "PublicTest"]

# Use a subset first so SVM is not too slow
train_sample = train_df.sample(n=8000, random_state=42)

X_train = np.stack(
    train_sample["pixels"].apply(lambda x: np.array(x.split(), dtype=np.float32) / 255.0)
)

y_train = train_sample["emotion"].values

X_val = np.stack(
    val_df["pixels"].apply(lambda x: np.array(x.split(), dtype=np.float32) / 255.0)
)

y_val = val_df["emotion"].values

print(X_train.shape, y_train.shape)
print(X_val.shape, y_val.shape)
```

```python
svm_model = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", LinearSVC(C=1.0, max_iter=5000))
])

svm_model.fit(X_train, y_train)
```

```python
val_preds = svm_model.predict(X_val)

acc = accuracy_score(y_val, val_preds)
print("SVM validation accuracy:", acc)

names = [EMOTIONS[i] for i in range(7)]
print(classification_report(y_val, val_preds, target_names=names))
```

```python
#The SVM is weak because it is probably using flattened 48×48 pixel values. That means it sees the face as a long vector of raw numbers, but it does not learn spatial patterns very well. It cannot understand local facial features like eyebrow shape, mouth curvature, wrinkles, or expression texture in the same way a CNN can.
#Your MobileNetV2, by contrast, learns hierarchical visual features, so it performs much better.
```

```python
from sklearn.metrics import classification_report

report = classification_report(
    y_val,
    val_preds,
    target_names=names,
    output_dict=True
)

report_df = pd.DataFrame(report).transpose()

FIGURES_DIR = os.path.join(PROJECT_ROOT, "figures")
report_path = os.path.join(FIGURES_DIR, "classification_report_svm.csv")

report_df.to_csv(report_path)
print("Saved:", report_path)
```

```python
print("The SVM baseline achieved 28.31% validation accuracy on the FER2013 PublicTest split, compared with 66.98% for MobileNetV2. This large performance gap suggests that raw pixel-based classical machine learning is insufficient for robust facial expression recognition. MobileNetV2 performed better because its convolutional layers can learn spatial and texture-based facial features, whereas the SVM baseline treats images as flattened vectors and loses much of the local structure.")
```
